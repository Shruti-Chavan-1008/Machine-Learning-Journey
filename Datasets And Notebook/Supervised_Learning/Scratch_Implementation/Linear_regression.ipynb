```python
import numpy as np
```

```python
#linear Regression

class LinearRegression:
    def __init__(self,learning_rate=0.01,n_iter=1000):
        self.bias=None
        self.weights=None
        self.lr=learning_rate
        self.n_iter=n_iter


    def fit(self , X, y): #X_train ,y_train
        m,n=X.shape #(number of samples,number of features)
        
        #step 1
        self.bias  = 0
        self.weights=np.zeros(n) #[0,0,...,0]
        for i in range(self.n_iter):
          #step2 -y pred value
          y_perd=self.bias + np.dot(X,self.weights)

          #step3 -cal griadent
          db=(1/m)*np.sum(y_perd - y)
          dw=(1/m)*np.dot(X.T,(y_perd - y))

          #step 4-- converage theroma param update
          self.bias= self.bias - self.lr * db
          self.weights=self.weights- self.lr * dw

        
        
    def perdict(self,X):
         y_perd=self.bias + np.dot(X,self.weights)
         return  y_perd
```

```python
X=np.array([[1],[2],[3],[4],[5]])
y=np.array([2,4,6,8,10])

model=LinearRegression()
model.fit(X,y)

y_pred=model.perdict(X)
print(y_pred)
```

```python
#linear regression with OLS (Ordinary leasrt square)

class LinearRegressionOLS:
    def __init__(self):
        self.bias=None
        self.weights=None

    def fit(self,X,y):
        m,n=X.shape

        X_b= np.c_[np.ones((m,1)),X]
        #normal equation
        theta=np.dot(np.linalg.inv(np.dot(X_b.T,X_b)), np.dot(X_b.T,y) )

        self.bias= theta[0]  
        self.weights=theta[1:]
        
    def perdict(self,X):
       y_perd=self.bias + np.dot(X,self.weights)
       return  y_perd
```

```python
X=np.array([[1],[2],[3],[4],[5]])
y=np.array([2,4,6,8,10])

model=LinearRegressionOLS()
model.fit(X,y)

y_pred=model.perdict(X)
print(y_pred)
```
