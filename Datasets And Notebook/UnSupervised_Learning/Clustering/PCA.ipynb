# Principal Component Analysis (PCA) :          
is an unsupervised dimensionality-reduction algorithm that transforms a dataset with many features into fewer new features while preserving as much of its variance as possible.

# Why do we use Principal Components (PC1, PC2)
We use principal components to reduce the number of features while keeping as much important variation in the data as possible.

# When do we use PCA
- Many features: You have 100 features and want to reduce them to 10.
- Highly correlated features: Several columns contain similar information.
- Visualization: You have many features and want to view patterns in 2D using PC1 and PC2.
- Noise reduction: You want to discard low-variance directions that mostly contain noise.
- Faster processing: Fewer features can make some downstream algorithms faster.

# What are PC1 and PC2?
PC1

First principal component
Captures the maximum possible variance along a single direction.



PC2

Second principal component
Is perpendicular to PC1 and captures the maximum remaining variance.



Remember: PC1 and PC2 are not necessarily original columns. They are new features calculated from combinations of the original features.
One caution: PCA does not always improve a model. If a low-variance feature is important for predicting the target, removing it can hurt performance




# code

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
```

```python
iris =load_iris()

X =pd.DataFrame(iris.data)
y=iris.target
```

```python
X.head()
```

```python
scaler=StandardScaler()
X_scaled=scaler.fit_transform(X)

X_scaled.shape
```

```python
#PCA 
from sklearn.decomposition import PCA

# 1. Initialize the PCA object
pca = PCA(n_components=2)

# 2. Fit the model and transform the data into a SEPARATE variable
X_pca = pca.fit_transform(X_scaled)

# 3. Now this will work perfectly!
print("explained variance ratio: ", pca.explained_variance_ratio_)
```

```python

X_pca.shape
```

```python
print(pca.components_)
```

```python
plt.figure(figsize=(8,6))

plt.scatter(
    X_pca[:,0],
    X_pca[:,1],
    c=y
)
plt.xlabel("PCA1")
plt.ylabel("PCA2")
plt.title("PCA for iris Datasets")
```
